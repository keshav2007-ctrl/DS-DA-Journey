```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# 1. Create simple training data (X = Houses Sizes, Y = House Prices)
X = np.array([[1000], [1500], [2000], [2500], [3000]])
y = np.array([200000, 280000, 360000, 440000, 520000])

# 2. Initialize and train the machine learning algorithm
model = LinearRegression()
model.fit(X, y)

# 3. Predict the price of a brand new 1800 sq ft home
predicted_price = model.predict(np.array([[1800]]) )
print(f"Predicted price for an 1800 sq ft house: ${predicted_price[0]:,.2f}")
```
